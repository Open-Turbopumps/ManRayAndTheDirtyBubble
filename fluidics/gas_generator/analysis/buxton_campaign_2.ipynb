$$ \delta m = \dot{m} t_1 + 1.5 \cdot \dot{m} t_2 $$

$$ \delta m = \dot{m} ( t_1 + 1.5 \cdot t_2) $$

$$ \dot{m} =\frac{\delta m}{( t_1 + 1.5 \cdot t_2)} $$

$$ T = (\dot{m}_f + \dot{m}_o)  \sqrt{\gamma R T} + A ( P_t - P_{amb} ) $$

## 2 - Analysis

```python
%%render

fuel_ctr = "C3H8O,2propanol"

ox_ctr = "N2O"

species_file = "config/n2o_ipa.yaml"
```

We can now define our Cantera combustion object.

```python
ctr = CombustionCantera(fuel = fuel_ctr,
                        oxidiser = ox_ctr,
                        species_file = species_file)
```

Now we put in our inputs

```python
%%render sci_not
 
thrust = 120 + 18 # N
D_t = 8e-3 # m
A_t = np.pi * (D_t /2 )**2 # m**2
P_amb = 1e5 # Pa
m_f = 35e-3 # kg/s
P_cc = 20e5 + 1e5
```

```python
m_o_guess = 0.062

mr_guess = m_o_guess / m_f

gas = ctr.get_thermo_prop(Pcc=P_cc, MR=mr_guess)

gamma = gas.gamma
R = gas.R
T = gas.t

T_t = T / ((gamma + 1)/2)
P_t = P_cc / ((gamma + 1)/2)**(gamma / (gamma - 1))

v_t = (gamma * R* T_t)**(1/2)

thrust_guess = (m_o_guess + m_f)*v_t + A_t * (P_t - P_amb)

error = (thrust_guess - thrust)/thrust

print(f"Stangation Temperature (K) {T:.2f}")
print(f"Throat Temperature (K): {T_t:.2f}")
print(f"Throat Pressure (Bar): {P_t/1e5:.2f}")
print(f"Error: {error*100:.2f} %")
```
