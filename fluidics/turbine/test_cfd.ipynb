```python
import pygmsh
import gmsh
import pandas as pd

import matplotlib.pyplot as plt
```

```python
geometry = pygmsh.geo.Geometry()
model = geometry.__enter__()
```

```python

p0 = model.add_point([0, 0, 0])
p1 = model.add_point([1, 0, 0])
p2 = model.add_point([1, 1, 0])
p3 = model.add_point([0, 1, 0])
l0 = model.add_line(p0, p1)
l1 =model.add_line(p1, p2)
l2 = model.add_line(p2, p3)
l3 = model.add_line(p3, p0)

loop = model.add_curve_loop([l0, l1, l2, l3])
surf = model.add_plane_surface(loop)

model.add_physical(entities=l0, label="inlet")
model.add_physical(entities=l1, label="outlet")
model.add_physical(entities=l2, label="top")
model.add_physical(entities=l3, label="bottom")
model.add_physical(entities=surf, label="Surface")
```

```python

mesh = geometry.generate_mesh(dim=2) 
gmsh.write("mesh.msh")
gmsh.clear()
geometry.__exit__()
```
